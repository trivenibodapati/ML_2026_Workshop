```python
#understanding the missing values and filling the missing values
data.isnull().sum()
```

```python
print(data['Age'].mean())
```

```python
data['Age'].isnull().sum()
```

```python
#we can fill the Age column missing values with mean
data['Age']=data['Age'].fillna(data['Age'].mean(),inplace=True)
data['Age'].isnull().sum()
```

```python
data['Embarked']=data['Embarked'].fillna('S',inplace=True)
```

```python
data.drop(columns=['Cabin'],inplace=True)
```

```python
data.isnull().sum()
```

```python
data.head(10)
```

```python
#get_dummies
pclass=pd.get_dummies(data['Pclass'])
pclass
```

```python
pclass=pd.get_dummies(data['Pclass'],drop_first=True)
pclass
```

```python
pclass=pd.get_dummies(data['Pclass'],drop_first=True,dtype=int)
pclass
```

```python
gender=pd.get_dummies(data['Sex'])
gender
```

```python
embarked=pd.get_dummies(data['Embarked'],drop_first=True,dtype=int)
embarked
```

```python
#concatenation of all rables
final_data = pd.concat([data,pclass,gender,embarked],axis=1)
final_data
```

```python
final_data.drop(columns=['PassengerId','Name','Pclass','Sex','Ticket','Embarked'],inplace=True)
final_data.head()
```
